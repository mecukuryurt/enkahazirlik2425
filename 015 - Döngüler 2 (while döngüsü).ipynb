````python
a = 0

while (a < 5):
    n1 = int(input("1. sınav notunuz:"))
    n2 = int(input("2. sınav notunuz:"))
    n3 = int(input("3. sınav notunuz:"))
    ort = (n1 + n2 + n3)/3
    print("Ortalamanız: ", ort)

    if (ort < 60):
        print("Kaldınız")
    else:
        print("Geçtiniz")
    a+=1
````
