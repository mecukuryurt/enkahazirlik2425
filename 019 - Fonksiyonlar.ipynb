````python
def toplama(a,b,c):
    print("Sayının toplamları:", a+b+c)
````

````python
def faktoriyel(sayi):
    if (sayi == 1 or sayi == 0):
        print("Faktöriyel:",1)
    else:
        res = 1 # result
        while (sayi >= 1):
            res *= sayi
            sayi -= 1
        print("Faktöriyel:",res)
````

````python
faktoriyel(0)
faktoriyel(7)
````

````python
def carp3(x):
    print("1.Fonksiyon Çalıştı")
    return x*3
def topla2(x):
    print("2.Fonksiyon Çalıştı")
    return x+2
def bol4(x):
    print("3.Fonksiyon Çalıştı")
    return x/4

print(bol4(topla2(carp3(5))))
# İç içe fonksiyon çağrıldığında ilk başta
# en içteki fonksiyon çağrılarak işlem yapılır
````

````python
# Fonksiyon parametreleri için Yıldızlı Parametre
# kullanılırsa fonksiyon çağrıldığında parametre
# olarak istenildiği kadar eleman verilebilir
def sayitopla(*a):
    sonuc = 0
    print("Parametreler:",a)
    for i in a:
        sonuc += i
    return sonuc

# Aşağıdaki durumda bütün parametreler fonksiyondaki
# a parametresinin içine geçecek
print(sayitopla(1,3,5,7,2,1,4,6,1))
````

````python
# Bir sosyal medya sitesinin kodlarında
# kullanıcı giriş yaptığında onu selamlayan
# bir kod olduğunu düşünelim
def girisyap(isim="isimsiz"):
    print("Merhaba {}! Hoşgeldiniz!".format(isim))

# Bu durumda isim="isimsiz" kod bloğu, hiçbir
# parametre girilmediğinde isim parametresinin
# alacağı değeri belirtmektedir.

girisyap("Fırat Sobutay") # isim="Fırat Sobutay"
girisyap("Abdülhey Çoban") # isim="Abdülhey Çoban"
girisyap() # isim="isimsiz"
````
