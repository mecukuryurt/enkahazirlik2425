# Tkinter Modülü Kullanımı

Bilgi: Bu ders notu tarafımca hazırlanmıştır, Erdinç Hoca derste yazdırmamıştır. Ayrıca bu notta sadece Tkinter modülüne temel bir şekilde değinilmekte olup detaya inilmemektedir. Daha fazla bilgi için https://docs.python.org/3/library/tkinter.html bkz.

Tkinter modülü, Python ile masaüstü uygulamaları yapmakta kullanılan bazı Python modüllerinden bir tanesidir. Bu modülün kullanımını anlamak için aşağıdaki örneğe göz atınız.

```python
# tkinter modülü, tk olarak içe aktarılır
import tkinter as tk
# tkinter modülünün içinden messagebox metot grubu, mb olarak aktarılır.
from tkinter import messagebox as mb

# tkinter modülü içindeki pencere oluşturmaya yarayan
# Tk sınıfı çağrılır ve root değişkeninin bir pencere
# objesi olması sağlanır. (bkz. Python'da Sınıflar)
root = tk.Tk()

# root adlı pencere objesinin metotlarından
# biri olan title metodu çağrılır ve pencerenin
# başlığı belirlenir.
root.title("Python Tkinter Denemesi")

# root objesinin metotlarından biri de geometry
# metodudur. Bu metot sayesinde pencerenin boyutu
# SATIRxSÜTUN şeklinde ayarlanır. 
root.geometry("500x100")

# tk modülünün içindeki Label (yazı) sınıfı çağrılır ve
# yazi1 adında bir Label objesi oluşturulur.
# İlk parametre olarak bu Label'ın bulunduğu pencere objesi belirtilir.
# Sonra istendiği takdirde text parametresiyle Label'ın metni, bg ile 
# arkaplan rengi, fg ile yazı rengi belirlenebilir.
yazi1 = tk.Label(root, text="Merhaba Tkinter!", bg="red", fg="yellow")
# Bu label objesi, kendisine ait grid metodu ile pencereye eklenir:
yazi1.grid(row=0, column=0)



sayac = 0
def butontiklandi():
    global sayac
    sayac += 1
    # Buton objesinin kendine ait config metodu ile daha sonradan
    # objenin özellikleri değiştirilebilir.
    buton1.config(text="Tıkla! {}".format(sayac))

    if sayac == 10:
        # messagebox'un showinfo metodu ile ekranda bir bilgi penceresi
        # gösterilir. İlk parametre başlığı, ikinci parametre yazıyı belirtir.
        # bkz. tkinter messagebox
        mb.showinfo("Tebrikler!", "10 kere tıkladınız!")


# tk modülünün içindeki Button sınıfı çağrılır ve buton1 adında bir
# Button objesi oluşturulur. İlk parametre olarak bu butonun bulnuacağı
# pencere objesi belirtilir. Daha sonra istenirse text parametresi ile
# buton üzerindeki metin, bg ile arkaplan rengi, fg ile yazı rengi, command
# ile butona tıklandığında yapılacak işlem belirlenir.
buton1 = tk.Button(root, text="Tıkla!", fg="white", bg="black", command=butontiklandi)
# Bu buton objesi, kendine ait grid metodu ile pencereye eklenir:
buton1.grid(row=0, column=1)

# tk modülünün içindeki Entry sınıfı çağrılır ve giris1 adında bir Entry objesi oluşturulur.
# İlk parametre olarak bu metin kutusunun bulunacağı pencere objesi belirtilir. 
# Daha sonra istenirse metin kutusunun yüksekliği ve genişliği, sırası ile
# height ve width parametreleri ile,  belirtilebilir.
giris1 = tk.Entry(root, width="20", fg="yellow", bg="blue")
# Bu metin kutusu objesi, kendine ait grid metodu ile pencereye eklenir:
giris1.grid(row=1, column=0)

def selamlabutonutiklandi():
    # Butona tıklandığında giris1 adlı metin kutusundan, kendine ait
    # get metodu ile içindeki metin alınır.
    metin = giris1.get()
    
    mb.showerror("Merhaba ben Volkan Konak", "Merhaba {}!".format(metin))

selamla = tk.Button(root, text="Selamla!", fg="red", bg="blue", command=selamlabutonutiklandi)
selamla.grid(row=1, column=1)

# root pencere objesi oluşturulduktan sonra kapatılana
# kadar ekranda durması sağlanır:
root.mainloop()
```
