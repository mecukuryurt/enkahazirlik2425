## Global ve Yerel Değişkenler

```python
a = 10 # Global Değişken
def fonk():
    print(a)
fonk()
print(a)
```

```python
def fonk1():
    b = 10 # fonk1 fonksiyonunun
           # yerel değişkenidir.
    print(b)

fonk1()
print(b)
```

Yukarıdaki örnekte ilk başta çıktı olarak <br>
10 yazılmakta fakat sonra hata vermektedir.<br>
Çünkü fonksiyonun içinde tanımlanan b değişkeni<br>
o fonksiyona aittir. İlk örnekteki a değişkeni <br>
global değişkendir, isteyen herkes bu değişkeni <br>
alıp kullanabilir fakat izin almadan değiştiremez <br>
Global değişken, kamu malına benzetilebilir.

Ayrıca hiçbir fonksiyon izin almadan global <br> 
bir değişkeni değiştiremez, fakat kullanabilir. <br>
Bir fonksiyon, global değiştiremez, aynı adda <br>
kendi yerel değişkenini oluşturur ve onu istediği <br>
gibi kullanır.

```python
c = 1923
def fonk2():
    c = 1453 # Global değişkeni değiştiremez.
             # Bu yüzden c adında bir yerel
             # değişken oluşturur.

    print(c) # Kendi yerel c değişkeni
fonk2()
print(c) # Global
```

```python
d = 1923
def fonk3():
    global d # Global olan değişkeni değiştirmek
             # için izin alıyoruz.

    d = 1453 # Global değişken, değiştirilir.
    print(d)

fonk3()
print(d)
```

Ayrıca global ve yerel değişkenler fonksiyonlar <br>
ile alakalıdır. for, while, if gibi işlemlerde <br>
yerel değişken olayı yoktur. 

```python
x = 5
while True:
    print("İlk:",x)
    x = 10
    print("İkinci:",x)
    break
print("Son:",x)
```

Sadece fonksiyonlarda yerel-global değişken olayı gözetilir.
