````python
def ebob(a:int, b:int):
    ebob = 1
    for i in range(1,min(a,b)+1):
        if a%i==0 and b%i==0:
            ebob = i
    
    return ebob

def ekok(a:int, b:int):
    # ebob*ekok = a*b
    return int((a*b) / ebob(a,b))

ebob(10,5)
````

### EBOB-EKOK kodunun açıklanması
İki sayının EBOB'unu hesaplamak için takip edeceğimiz birden fazla algoritma bulunmaktadır.<br>
Algoritmalardan bir tanesi şudur: <br>
- Sayıların en küçüğü seçilir.
- 1'den bu sayıya kadar olan bütün sayılar dolaşılır.
- Bu dolaşılan sayılar arasından bir sayı, girilen iki sayıya da bölünüyorsa bu sayı bir değişkene atanır.
- Sayılar küçükten büyüğe doğru dolaşılacağı için değişkene en son atanacak sayı kalacaktır ve bu sayı
  en büyük bölen olacaktır.
- Ayrıca a*b=ebob(a,b)*ekok(a,b) olduğu için ebob bulununca ekok da matematik işlemi ile bulunur.
<br>

Fonksiyon tanımlanır, a ve b parametrelerinin integer girilmesi gerektiği belirtilir. ebob değişkeni oluşturulur:

```python
def ebob(a:int, b:int):
    ebob = 1
```
<br>
Küçük sayı ile büyük sayı arasındaki hiçbir sayı, bu iki sayıya tam bölünemeyeceği için en küçük sayıda<br>
durmamız yeterlidir. range komutu, girdiğimiz sayıyı almaz, bu sayıdan bir önceki sayıda durur. Bunun için <br>
küçük sayıyı bulduktan sonra 1 ekliyoruz:

```python
    for i in range(1,min(a,b)+1):
```

i değişkeni, 1'den küçük sayıya kadar bütün sayıları dolaşacak. Eğer i sayısı, girilen a ve b sayılarının <br>
ikisine de bölünüyorsa, bu iki sayının ortak bölenidir. Sayılar küçükten büyüğe dolaşılacağı için başka bir <br>
ortak bölen bulunduğunda bu, kesinlikle önceki ortak bölenden büyük olacaktır. Aynı değişkene atamamızda sakınca <br>
yoktur:

```python
        if a%i==0 and b%i==0:
            ebob = i
```

En sonunda ebob değişkenini geri döndürüyoruz:
```python
    return ebob
```
Ayrıca $a*b = ebob(a,b)*ekok(a,b)$ eşitliğinden yararlanarak <br>
$ekok(a,b) = (a*b)\over ebob(a,b)$ eşitliğine ulaşırız. <br>
Böylece ekok değeri aşağıdaki gibi bulunur:

```python
def ekok(a:int, b:int):
    # ebob*ekok = a*b
    return int((a*b) / ebob(a,b))
```
EBOB, EKOK işlemlerinde sonucun ondalıklı sayı çıkma olasılığı yoktur. <br>
Python'daki bölme işlemi ondalıklı sonuç verdiğinden bu değeri integer değerine çeviriyoruz.
