````python
# Ağaç çizen program:
length = 10
for i in range(1, length+1):
    print(" "*(length-i) + "* "*i)
````

### Ağaç kodunun açıklanması
Ağacın uzunluğunu belirliyoruz:
```python
length = 10
```
Daha sonra ağacın her bir katı için işlem yapacağız.<br>
Döngüyü başlatıyoruz:
```python
for i in range(1, length+1):
    print(" "*(length-i) + "* "*i)
```
Ağacı yaparken basamaklardaki yıldız(*) karakterlerini ortalamamız gerekli.<br>
Bunu yaparken her basamakta, kendinden sonraki basamak sayısı kadar boşluk bırakmalıyız. <br>
Bu boşluk sayısı 1. basamakta 9, 10. basamakta 0 olacaktır. Bunu aşağıdaki ifade yapıyor: <br> 
```python
" "*(length-i)
```
Ayrıca her basamakta, bulunulan basamağın numarası kadar yıldız(*) olacak ve her yıldız arası 1 birim boşluk olacak. <br>
Bunu gerçekleştiren kod parçası ise şudur: <br>
```python
"* "*i
```
