## Soru 3: Dairenin Alanı

```python
r = int(input("Dairenin yarıçapını giriniz: "))

def dairealanbul(yaricap):
    pi = 3.14
    return pi * (yaricap**2)

res = dairealanbul(r)
print("Dairenin Yarıçapı: {}\nDairenin Alanı: {}\n".format(r,res))
```

## Soru 4: Para İşlemleri

```python
print("Banka programına hoşgeldiniz.")
bakiye = 1000
while True:
    print("#"*10)
    print("Lütfen aşağıdan bir işlem seçiniz:")
    print("1 - Bakiye Sorgulama\n2 - Para Yatırma\n3 - Para Çekme\nq - Çıkış")
    islem = input("Lütfen işlem seçiniz: ")
    if islem == "q" or islem=="Q":
        print("Yine bekleriz...")
        break
    elif islem == "1":
        print("Bakiyeniz: {} TL'dir".format(bakiye))
    elif islem == "2":
        yatirilacakpara = int(input("Kaç lira yatırmak istiyorsunuz: "))
        bakiye += yatirilacakpara
        print("Paranız başarı ile yatırıldı.")
    elif islem == "3":
        cekilecekpara = int(input("Kaç lira para çekmek istiyorsunuz: "))
        bakiye -= cekilecekpara
        print("Paranız başarı ile çekilmiştir.")
```

## Soru 5: Sayının Okunuşu

```python
sayi = input("İki basamaklı bir tam sayı giriniz: ")

def okunusbul(sayi):
    onlarbas  = ["", "on", "yirmi", "otuz", "kırk", "elli", "altmış", "yetmiş", "seksen", "doksan"]
    birlerbas = ["", "bir", "iki", "üç", "dört", "beş", "altı", "yedi", "sekiz", "dokuz"]

    eksi = ""
    if sayi[0] == "-":
        eksi = "eksi "
    else:
        sayi = " {}".format(sayi)
    
    okunus = "{}{} {}".format(eksi, onlarbas[int(sayi[1])], birlerbas[int(sayi[2])])
    return okunus

okunus = okunusbul(sayi)
print("Okunuşu: {}".format(okunus))
```
