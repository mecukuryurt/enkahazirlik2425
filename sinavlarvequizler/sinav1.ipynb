# 1. Soru

### Geçti Kaldı Sorusu

```python
pt1 = int(input("1. Yazılı Notunu Giriniz: "))
pt2 = int(input("2. Yazılı Notunu Giriniz: "))
pt3 = int(input("3. Yazılı Notunu Giriniz: "))

ortalama = (pt1 + pt2 + pt3) / 3

if ortalama > 60:
    print("GEÇTİ")

else:
    print("KALDI")
```

# 2. Soru

### Tek Çift Sorusu

```python
num = int(input("Bir sayı giriniz: "))

if num % 2 == 1:
    print("Sayı TEK")

else:
    print("Sayı ÇİFT")
```

# 3. Soru

### En Büyük Sayı Sorusu

```python
a = int(input("1. sayıyı giriniz: "))
b = int(input("2. sayıyı giriniz: "))
c = int(input("3. sayıyı giriniz: "))

print()
if a >= b and a >= c:
    print(a)
elif b >= a and b >= c:
    print(b)
elif c >= a and c >= b:
    print(c)
```

# 5. Soru

### Variable Swap Sorusu

```python
a = int(input("1. Sayıyı girin: "))
b = int(input("2. Sayıyı girin: "))

print("\na={} ve b={}\nSwap işlemi yapılacak".format(a,b))
a,b = b,a
print("Swap işlemi yapıldı!\na={} b={}".format(a,b))
```
