## Problem 1: Beden Kitle İndeksi

```python
boy = float(input("Lütfen boyunuzu metre cinsinden girin: "))
kilo = float(input("Lütfen kilonuzu kilogram cinsinden girin: "))

if (boy <= 0 or kilo <= 0):
    print("Yanlış girdiniz!")

else:
    bki =kilo / (boy * boy)
    print("Beden kitle indeksiniz: {}".format(bki)) 
```

## Problem 2: Vize, Final Notu Hesaplama

```python
v1 = int(input("1. Vize Notunuzu Girin: "))
v2 = int(input("2. Vize Notunuzu Girin: "))
ff = int(input("Final Notunuzu Girin  : "))

notvalidval = (v1 > 100 or v1 < 0) or (v2 > 100 or v2 < 0) or (v2 > 100 or v2 < 0)
if notvalidval:
    print("Yanlış not girdiniz!")

sn = v1*0.3 + v2*0.3 + ff*0.4

if   sn >= 90: res = "AA"
elif sn >= 85: res = "BA"
elif sn >= 80: res = "BB"
elif sn >= 75: res = "CB"
elif sn >= 70: res = "CC"
elif sn >= 65: res = "DC"
elif sn >= 60: res = "DD"
elif sn >= 55: res = "FD"
elif sn <  55: res = "FF"

if not notvalidval:
    print("Notunuz: {}. Harf Notunuz: {}".format(sn, res))
```

## Problem 3: Geometrik Şekiller

```python
objt = input("Hangi şekil için hesap yapacaksınız (Üçgen, Dörtgen): ")

if objt == "Dörtgen":
    knr = []
    for i in range(4): knr.append(int(input("{}. kenarın uzunluğunu giriniz: ".format(i+1))))
    if knr[0] == knr[1] == knr[2] == knr[3]:
        print("Girdiğiniz şekil KARE'dir")
    elif knr[0] == knr[2] and knr[1] == knr[3]:
        print("Girdiğiniz şekil DİKDÖRTGEN'dir")
    else:
        print("Girdiğiniz şekil herhangi bir dörtgendir")

elif objt == "Üçgen":
    knr = []
    for i in range(3): knr.append(int(input("{}. kenarın uzunluğunu giriniz: ".format(i+1))))

    yanlisucgenmi1 = abs(knr[0] - knr[1]) >= knr[2] # Kuralı sağlamayan kenar var mı?
    yanlisucgenmi2 = abs(knr[1] - knr[2]) >= knr[0]
    yanlisucgenmi3 = abs(knr[0] - knr[2]) >= knr[1]
    
    if yanlisucgenmi1 or yanlisucgenmi2 or yanlisucgenmi3:
        print("Bu bir üçgen değildir!")
    elif knr[0] == knr[1] == knr[2]:
        print("Bu üçgen EŞKENARDIR")
    elif knr[0] == knr[1] or knr[0] == knr[2] or knr[1] == knr[2]:
        print("Bu üçgen İKİZKENARDIR")
    else:
        print("Bu sıradan bir üçgendir.")

else:
    print("Yanlış girdiniz!")
```
