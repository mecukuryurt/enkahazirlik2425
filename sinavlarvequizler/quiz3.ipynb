```python
def prob1(s1, s2): # Girilen 2 sayı arasındaki cift sayilarin ortalamasi
    nc = 0 # Number Count
    topl = 0
    if s2 < s1: s2,s1=s1,s2
    for i in range(s1,s2+1):
        if i % 2 == 0:
            nc += 1
            topl += i
    ort = topl / nc
    return ort

a = int(input("1. Sayiyi Girin: "))
b = int(input("2. Sayiyi Girin: "))
print("Ortalama: ", prob1(a,b))
```

```python
def pisagormu(a,b):
    c = (a*a + b*b)**(0.5)
    return (c % 1 == 0, int(c))
for a in range(1,101):
    for b in range(a+1,101):
        res = pisagormu(a,b)
        if res[0]:
            print("Pisagor Üçgeni: ", a,b,res[1])
```

```python
def sekiltipi(k1,k2,k3,k4):
    if k1 == k2 == k3 == k4: return "Kare"
    elif k1==k3 and k2==k4: return "Dikdörtgen"
    else: return "Diğer Dörtgen"

inputs = [0,0,0,0]
for i in range(4):
    inputs[i] = int(input("{}. kenarı giriniz: ".format(i+1)))

print("Şeklin Tipi:", sekiltipi(inputs[0],inputs[1],inputs[2],inputs[3]))
```

```python
sehirler = ["Ankara", "Bursa", "Ceyhan", "Çanakkale", "Düzce", "Erzurum", "Fatsa", "Giresun", "Ğ ile yok: Elazığ", "Hakkari", "Isparta", "İstanbul",
            "Jale", "Karaman", "Lüleburgaz", "Manisa", "Niğde", "Ordu", "Ömerli", "Pendik", "Rize", "Sivas", "Şanlıurfa", "Trabzon", "Uşak",
            "Üsküdar", "Van", "Yozgat", "Zonguldak"]
def sehirlistele(harf):
    for sehir in sehirler:
        if sehir[0] == harf:
            return sehir

def upper(harf):
    if harf == "i": return "İ"
    else: return harf.upper()

# print(dir("asd"))
isim = input("İsminiz: ")
for harf in isim:
    print(sehirlistele(upper(harf)))

# abcçdefgğhıijklmnoöprsştuüvyz
```
