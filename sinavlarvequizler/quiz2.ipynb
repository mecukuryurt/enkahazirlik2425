# Problem 1: Mükemmel Sayı

```python
sayi = int(input("Bir sayı giriniz: "))

toplam = 0
for i in range(1,sayi):
    if sayi % i == 0: # Bölenidir
        toplam += i

if toplam == sayi:
    print("Mükemmel Sayıdır")

else:
    print("Mükemmel Sayı Değildir")
    
# 6, 28, 496, 8128
```

# Problem 2: Armstrong Sayısı

```python
sayi = input("Bir sayı giriniz: ")

toplam = 0
uzn = len(sayi)

for c in sayi:
    toplam += int(c) ** uzn

if toplam == int(sayi):
    print("Armstrong Sayısıdır")
else:
    print("Armstrong Sayısı Değildir")
     
# 153, 370, 371, 407, 1634, 8208, 9474
```

# Problem 3: Toplam Programı

```python
toplam = 0

while True:
    sayi = input("Bir Sayı Giriniz (çıkmak için q): ")

    if sayi == "q":
        print("Çıkılıyor")
        break
    else:
        toplam += int(sayi)

print("Sayıların Toplamı: ", toplam)
```

# Problem 4: 3'e Bölünen Sayılar

```python
for i in range(1,100):
    if i % 3 != 0: 
        continue
    print(i)
```
