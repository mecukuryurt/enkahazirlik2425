### if-elif-else kod yapısı:
```python
if (koşul):
    yapılacak işlemler
    
elif (koşul):
    yapılacak işlemler
    
else:
    yapılacak işlemler
```

````python
islem = int(input("İşlem Seçiniz (1,2,3): "))
if islem == 1:
    print("1. işlemi seçtiniz.")
elif islem == 2:
    print("2. işlemi seçtiniz.")
elif islem == 3:
    print("3. işlemi seçtiniz.")
else:
    print("Geçersiz işlem seçtiniz!")
````

````python
print("Düşük bütçeli hesap makinesi programıma hoşgeldiniz!")
islem = input("İşlemi giriniz (+-*/):")
a = float(input("1. Sayıyı Giriniz: "))
b = float(input("2. Sayıyı Giriniz: "))

if islem == "+":
    print("{}+{}={}".format(a,b,a+b))
elif islem == "-":
    print("{}-{}={}".format(a,b,a-b))
elif islem == "*":
    print("{}*{}={}".format(a,b,a*b))
elif islem == "/":
    print("{}/{}={}".format(a,b,a/b))
````
