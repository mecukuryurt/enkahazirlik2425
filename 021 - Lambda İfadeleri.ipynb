## Lambda İfadeleri

```python
fonk_adi = lambda param1, param2: donusdegeri
```

````python
liste = [1,2,3,4,5]
liste1 = list()

for i in liste:
    liste1.append(i)

print(liste1)
````

````python
a = [i for i in liste]
print(a)
````

Yukarıdaki gibi listeye ekleme yapma işlemini kısaltabildiğimiz gibi, <br>
geri değer döndüren bazı fonksiyonlarını da kısaltabiliriz.

````python
def carp2(x):
    return x*2
print(carp2(12))
````

````python
# carp2 fonksiyonunu 
# aşağıdaki gibi 
# kısaltabiliriz.
carp2 = lambda x: x*2
# Bu ifade, üst hücredeki
# fonksiyon tanımı ile aynı
# şeyleri yapar.
print(carp2(12))
````

````python
topla = lambda a,b,c: a+b+c
topla(3,4,5)
````

````python
terscevir = lambda s: s[::-1]
terscevir("aynüd abahrem")
````

````python
ciftmi = lambda num: num%2==0
print(ciftmi(8))
print(ciftmi(9))
````

````python
# Başka bir kullanım:
selamla = lambda ad, soyad: print("Hoşgeldiniz {}, {}!".format(ad,soyad))
selamla("Polat","Alemdar")
````
