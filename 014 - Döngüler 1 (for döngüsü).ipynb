````python
# Listedeki elemanların toplamını bulan program
liste = [1,2,3,4,5,6,7]
toplam = 0
for eleman in liste:
    toplam += eleman

print("Toplam: ", toplam)
````

````python
# Listedeki çift ve tek elemanların ayrı ayrı sayısını bulan program
liste = [1,2,3,4,5,6,7]
ciftsayisi = 0
teksayisi = 0

for eleman in liste:
    if eleman % 2 == 0: # çifttir
        ciftsayisi += 1
    else:
        teksayisi += 1

print("Çiftlerin Sayısı: ", ciftsayisi)
print("Teklerin Sayısı : ", teksayisi)
````

````python
# Listedeki çift ve tek elemanların ayrı ayrı toplamını bulan program
liste = [1,2,3,4,5,6,7]
cifttoplam = 0
tektoplam = 0

for eleman in liste:
    if eleman % 2 == 0: # çifttir
        cifttoplam += eleman
    else:
        tektoplam += eleman

print("Çiftlerin Toplamı: ", cifttoplam)
print("Teklerin Toplamı : ", tektoplam)
````
